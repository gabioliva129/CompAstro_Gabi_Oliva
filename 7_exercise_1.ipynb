GAUSSIAN ELIMINATION
--> SOLVING MATRICES

2w + x +  4y + z  = -4 \
3w + 4x - y  - z  = 3 \
w  + 4x + y +  5z = 9 \
2w + 2x + y +  3z = 7


when making matrices, you need to use 2D arrays

each value of the overall array has to be the hotizontal component of the matrix.

this system of equaiton can be solved with the following equation:
A * x_vector = v_vector

where A is the matrix, x_vector is what we're solving for (w, x, y, z), and y is the results

```python
import numpy as np

A = np.array([[2,1,4,1], [3,4,-1,-1], [1,4,1,5], [2,2,1,3]], float)
v_vector = np.array([-4,3,9,7], float)
x_vector = len(v_vector)
```

```python
for m in range(x_vector):
    print(A[m, :])
    v_vector[m] =  v_vector[m] / A[m,m]
    A[m,:]= A[m,:] / A[m,m]
    print(A[m, :])
    print(v_vector)

    A[m,:] = -A[m,0]*A[m,:]
```
